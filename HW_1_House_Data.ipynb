## Problem 5 mean/min/max/variance and correlations

```python

df = pd.read_csv("kc_house_data.csv")

features = df.drop(columns=["id", "date", "zipcode", "price"])

stats = pd.DataFrame({
    "mean": features.mean(),
    "min": features.min(),
    "max": features.max(),
    "variance": features.var()  
}).sort_index()

stats

```

```python
lowest_avg = stats["mean"].idxmin(), stats["mean"].min()
highest_avg = stats["mean"].idxmax(), stats["mean"].max()

lowest_var = stats["variance"].idxmin(), stats["variance"].min()
highest_var = stats["variance"].idxmax(), stats["variance"].max()

print('Lowest average:',lowest_avg)
print('Highest average:', highest_avg)
print('Lowest variance:',lowest_var)
print('Highest variance:', highest_var)
```

```python
correlations = df[features.columns].corrwith(df["price"])

corr_table = correlations.to_frame('Correlation: feature and price')

corr_table
```

```python
positive_corr = correlations[correlations > 0]
negative_corr = correlations[correlations < 0]

strongest_corr = correlations.idxmax(), correlations.max()

print("Positively correlated features:", list(positive_corr.index))
print("Highest positive correlation:", strongest_corr)

print('Negatively correlated features:', list(negative_corr.index))
```
